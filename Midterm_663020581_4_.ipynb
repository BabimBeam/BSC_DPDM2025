```python
import pandas as pd
import numpy as np

# แปลงวันที่ให้เป็น datetime object เพื่อความชัวร์
hii_daily_rain['measure_datetime'] = pd.to_datetime(hii_daily_rain['measure_datetime'])

# 1. ดึงรายชื่อสถานีเป้าหมาย (Nodes)
target_stations = Stations_model_HII['Node'].unique()

# นับจำนวนก่อนกรอง
count_before = len(hii_daily_rain)
stations_before = hii_daily_rain['station_code'].nunique()

# กรองข้อมูล
df_filtered = hii_daily_rain[hii_daily_rain['station_code'].isin(target_stations)].copy()

# นับจำนวนหลังกรอง
count_after = len(df_filtered)
stations_after = df_filtered['station_code'].nunique()

print(f"--- ผลการกรองสถานี ---")
print(f"จำนวน Record ก่อนกรอง: {count_before} (จาก {stations_before} สถานี)")
print(f"จำนวน Record หลังกรอง: {count_after} (เหลือ {stations_after} สถานี)")
```

#3. ระบุและจัดการข้อมูลสูญหาย (Missing Data)
การหา "Record ที่หายไป" (Missing Records) ในข้อมูล Time Series ต้องสร้างช่วงเวลา (Date Range) ที่สมบูรณ์ขึ้นมาก่อน เพื่อดูว่าวันไหนหายไป

```python
# สร้างช่วงเวลาตั้งแต่วันแรกถึงวันสุดท้ายที่มีข้อมูล
min_date = df_filtered['measure_datetime'].min()
max_date = df_filtered['measure_datetime'].max()
full_date_range = pd.date_range(start=min_date, end=max_date, freq='D')

# สร้าง MultiIndex ของ (ทุกสถานี x ทุกวัน)
all_stations = df_filtered['station_code'].unique()
mux = pd.MultiIndex.from_product([all_stations, full_date_range], names=['station_code', 'measure_datetime'])

# Reindex ข้อมูลเดิมเข้ากับ Index ใหม่ (วันที่ที่ไม่มีข้อมูลจะเป็น NaN)
df_full = df_filtered.set_index(['station_code', 'measure_datetime']).reindex(mux).reset_index()

# คำนวณจำนวนและสัดส่วนข้อมูลที่หายไป (นับจากคอลัมน์ data ที่เป็น NaN)
missing_stats = df_full.groupby('station_code')['data'].apply(lambda x: x.isnull().sum())
total_records = len(full_date_range)
missing_percent = (missing_stats / total_records) * 100

# แสดงตัวอย่าง
print("--- สถิติข้อมูลสูญหายรายสถานี (บางส่วน) ---")
print(missing_percent.head())
```

#4. คัดสถานีที่มีข้อมูลสูญหาย $\ge 20\%$ ออก

```python
# หา list รายชื่อสถานีที่ข้อมูลหายมากกว่าหรือเท่ากับ 20%
stations_to_drop = missing_percent[missing_percent >= 20].index
stations_to_keep = missing_percent[missing_percent < 20].index

# แสดงผลการคัดออก
print(f"--- การคัดเลือกสถานี ---")
print(f"จำนวนสถานีที่ถูกคัดออก: {len(stations_to_drop)}")
print(f"รายชื่อสถานีที่ถูกคัดออก: {list(stations_to_drop)}")
print(f"จำนวนสถานีที่เหลืออยู่: {len(stations_to_keep)}")

# กรอง DataFrame ให้เหลือเฉพาะสถานีที่ผ่านเกณฑ์
df_clean = df_full[df_full['station_code'].isin(stations_to_keep)].copy()
```

#5. เติมข้อมูลที่หายไปด้วยค่า Median (Day of Year)
เราจะใช้ค่ามัธยฐานของ "วันที่ของปี" (เช่น ค่าเฉลี่ยของทุกวันที่ 1 มกราคม ของสถานีนั้นๆ) มาเติมลงในช่องว่าง

```python
# สร้างคอลัมน์ Day of Year (1-366)
df_clean['doy'] = df_clean['measure_datetime'].dt.dayofyear

# คำนวณ Median แยกตาม สถานี และ Day of Year
# transform จะช่วยกระจายค่ากลับไปให้เท่ากับจำนวนแถวเดิม เพื่อให้ง่ายต่อการ fillna
median_values = df_clean.groupby(['station_code', 'doy'])['data'].transform('median')

# เติมข้อมูล: ถ้า data ว่าง ให้ใช้ค่าจาก median_values, ถ้ามีค่าอยู่แล้วให้ใช้ค่าเดิม
df_clean['data'] = df_clean['data'].fillna(median_values)

# กรณีที่ Median ก็ยังเป็น NaN (เช่น วันที่ 29 ก.พ. ไม่เคยมีฝนเลยตลอดประวัติ) อาจเติมด้วย 0 หรือ Forward Fill
df_clean['data'] = df_clean['data'].fillna(0)

# ตรวจสอบว่าไม่มี record หายอีก
remaining_nulls = df_clean['data'].isnull().sum()
print(f"--- ตรวจสอบการเติมข้อมูล ---")
print(f"จำนวนค่าว่างที่เหลืออยู่: {remaining_nulls}")
```

#6.อภิปรายเหตุผล: ทำไมจึงใช้ Median แทน Mean?
ในการจัดการข้อมูลทางอุทกวิทยา (Hydrology) หรือปริมาณน้ำฝน เหตุผลหลักที่นิยมใช้ Median (ค่ามัธยฐาน) มากกว่า Mean (ค่าเฉลี่ย) มีดังนี้:

* การกระจายตัวของข้อมูล (Skewed Distribution): ข้อมูลปริมาณน้ำฝนมักมีการแจกแจงแบบเบ้ขวา (Right-skewed) คือมีวันที่ฝนไม่ตก (0 mm) จำนวนมาก และมีวันที่ฝนตกหนักมากๆ (Extreme values) จำนวนน้อย

* ค่า Mean จะถูกดึงให้สูงขึ้นอย่างมากโดยค่าฝนตกหนักเพียงไม่กี่วัน ทำให้ค่าที่ได้ไม่สะท้อน "เหตุการณ์ปกติ"

* ค่า Median จะอยู่ตรงกลางของข้อมูลจริงๆ และได้รับผลกระทบจากค่าสุดโต่ง (Outliers) น้อยกว่า จึงเป็นตัวแทนของปริมาณฝนในวันที่ "ปกติ" ได้ดีกว่า

* ในกรณีที่มีค่าว่างไม่เยอะมาก จึงใช้คำที่พบบ่อยมากที่สุดในการจัดการค่าว่าง

## Handle Missing Daily Records (2025)

### Subtask:
Create a complete date range for the entire year 2025. Then, reindex the filtered API_2025 DataFrame with a MultiIndex of all filtered stations and this full date range to explicitly show any missing daily records.


```python
import pandas as pd

# 1. Extract the minimum and maximum dates from the measure_datetime column
min_date_api = API_2025_filtered['measure_datetime'].min()
max_date_api = API_2025_filtered['measure_datetime'].max()

# 2. Generate a complete date range
full_date_range_api = pd.date_range(start=min_date_api, end=max_date_api, freq='D')

# 3. Identify all unique station_code values
all_stations_api = API_2025_filtered['station_code'].unique()

# 4. Create a MultiIndex of (all stations x every day)
mux_api = pd.MultiIndex.from_product([all_stations_api, full_date_range_api], names=['station_code', 'measure_datetime'])

# 5. Set the station_code and measure_datetime columns as the index and reindex it with the newly created MultiIndex
# 6. Reset the index of the reindexed DataFrame to convert the MultiIndex back into columns.
# 7. Store the resulting DataFrame in a new variable, API_2025_full
API_2025_full = API_2025_filtered.set_index(['station_code', 'measure_datetime']).reindex(mux_api).reset_index()

print("Successfully created API_2025_full with a complete date range for each station.")
print("First 5 rows of API_2025_full:")
print(API_2025_full.head())
print("Last 5 rows of API_2025_full:")
print(API_2025_full.tail())
print(f"Shape of API_2025_full: {API_2025_full.shape}")
```

Data Interpolation (Time-Series)" หรือการขยายตารางข้อมูลเพื่อให้ "ทุกสถานีมีข้อมูลครบทุกวัน

**รายละเอียดการทำงานทีละขั้นตอน**
1. หาช่วงเวลา (Min/Max Date): หาจุดเริ่มต้นและจุดสิ้นสุดของข้อมูลทั้งหมดในเซตนี้
2. สร้างปฏิทินอ้างอิง (pd.date_range): สร้างรายการวันที่เรียงกันแบบ "รายวัน" (freq='D') ตั้งแต่วันแรกจนถึงวันสุดท้ายแบบไม่ข้ามแม้แต่วันเดียว
3. หาจำนวนสถานีทั้งหมด: ดึงรายชื่อสถานีที่ไม่ซ้ำกันออกมา
4. สร้างตารางเปล่าแบบ "คูณไขว้" (pd.MultiIndex):
* ขั้นตอนนี้คือหัวใจสำคัญครับ มันจะเอา (รายชื่อทุกสถานี) × (ทุกวันที่ในปฏิทิน)
* สมมติมี 10 สถานี และมี 180 วัน คุณจะได้ตารางที่มี $10 \times 180 = 1,800$ แถวแน่นอน
5. การ reindex
* นำตารางจริงของคุณไปวางทับบนตารางเปล่าที่เพิ่งสร้าง
* แถวไหนที่มีข้อมูลอยู่แล้ว (มีค่าฝน) ข้อมูลก็จะยังอยู่แถวไหนที่ในตารางจริงไม่มี (เช่น วันที่ 5 สถานี A หายไป): Pandas จะสร้างแถวของวันที่ 5 สถานี A ขึ้นมาให้ใหม่โดยอัตโนมัติ แต่ค่าปริมาณฝนจะกลายเป็น NaN (ค่าว่าง)

## Drop Stations with Excessive Missing Data (2025)




```python
# 1. Calculate the number of missing `data` values for each `station_code`
missing_data_count_api = API_2025_full.groupby('station_code')['data'].apply(lambda x: x.isnull().sum())

# 2. Determine the total number of days in the `full_date_range_api`
total_days_api = len(full_date_range_api)

# 3. Calculate the percentage of missing `data` for each station
missing_percent_api = (missing_data_count_api / total_days_api) * 100

# 4. Identify stations to drop (>= 20% missing)
stations_to_drop_api = missing_percent_api[missing_percent_api >= 20].index.tolist()

# 5. Identify stations to keep (< 20% missing)
stations_to_keep_api = missing_percent_api[missing_percent_api < 20].index.tolist()

# 6. Print the results
print(f"--- API 2025 Station Filtering ---")
print(f"Total stations: {len(all_stations_api)}")
print(f"Number of stations to be dropped (>= 20% missing): {len(stations_to_drop_api)}")
print(f"Stations to be dropped: {stations_to_drop_api}")
print(f"Number of stations to be kept (< 20% missing): {len(stations_to_keep_api)}")
print(f"Stations to be kept: {stations_to_keep_api}")

# 7. Filter the DataFrame to keep only the desired stations
API_2025_clean = API_2025_full[API_2025_full['station_code'].isin(stations_to_keep_api)].copy()

print("\nFirst 5 rows of API_2025_clean after filtering:")
print(API_2025_clean.head())
print(f"Shape of API_2025_clean: {API_2025_clean.shape}")
```

```python
API_2025_clean['month'] = API_2025_clean['measure_datetime'].dt.month
API_2025_clean['year'] = API_2025_clean['measure_datetime'].dt.year

API_2025_monthly_rainfall = API_2025_clean.groupby(['station_code', 'year', 'month'])['data'].sum().reset_index()

API_2025_monthly_rainfall = API_2025_monthly_rainfall.rename(columns={'station_code': 'Node', 'data': 'rain'})

API_2025_monthly_rainfall = API_2025_monthly_rainfall[['Node', 'rain', 'month', 'year']]

print("Monthly rainfall for 2025 calculated and formatted successfully.")
print("First 5 rows of API_2025_monthly_rainfall:")
print(API_2025_monthly_rainfall.head())
print(f"Shape of API_2025_monthly_rainfall: {API_2025_monthly_rainfall.shape}")
```
